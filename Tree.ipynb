```python
class Node:
  def __init__(self, data):
    self.data = data
    self.left = None
    self.right = None

class BinarySearchTree:
  def __init__(self):
    self.root = None
  def insert(self, root, data):
    if root is None:
      return Node(data)
    if data < root.data:
      root.left = self.insert(root.left, data)
    else:
      root.right = self.insert(root.right, data)
    return root

bst = BinarySearchTree()
root = None
data_list = [50, 30, 70, 20, 40, 60, 80]
for value in data_list:
  root = bst.insert(root, value)

print("Hasil inorder traversal:")
inorder(root)
```

```python
class Node:
    def __init__(self, data):
        self.data = data      # pendefinisian data node utama
        self.left = None      # pendefinisian root kiri
        self.right = None     # pendefinisian root kanan

# Membuat tree secara manual
root = Node(19)              # jumlah huruf dari nama "Albani Hafiz Innopha"
root.left = Node(8)          # jumlah huruf vokal
root.right = Node(24)        # 2 digit terakhir npm (contoh)
root.right.right = Node(5)   # bulan lahir (contoh Mei)
root.left.left = Node(12)    # tanggal lahir (contoh 12)

# Fungsi inorder traversal
def inorder(node):
    if node:
        inorder(node.left)           # traversal dari node kiri
        print(node.data, end=' ')    # menampilkan node utama
        inorder(node.right)          # traversal dari node kanan

print("Hasil inorder traversal dari binary tree:")
inorder(root)    # pemanggilan fungsi traversal inorder
```

```python
class TreeNode:
    def __init__(self, nilai):
        self.data = nilai          # pendefinisian data utama
        self.left = None           # pendefinisian cabang kiri
        self.right = None          # pendefinisian cabang kanan

class BST:
    def __init__(self):
        self.root = None           # pendefinisian root awal

    def tambah_node(self, root, nilai):
        if root is None:
            return TreeNode(nilai)   # mengembalikan node jika masih kosong

        if nilai < root.data:
            root.left = self.tambah_node(root.left, nilai)   # masuk ke node kiri
        else:
            root.right = self.tambah_node(root.right, nilai) # masuk ke node kanan

        return root                 # mengembalikan root

# Membuat objek BST
tree = BST()

# Root awal kosong
akar = None

# Data yang akan dimasukkan
angka = [40, 10, 10, 70, 65, 12, 7]

# Perulangan untuk menambahkan data
for item in angka:
    akar = tree.tambah_node(akar, item)

# Traversal inorder
def inorder(node):
    if node:
        inorder(node.left)              # menampilkan bagian kiri
        print(node.data, end=' ')       # menampilkan node utama
        inorder(node.right)             # menampilkan bagian kanan

# Traversal preorder
def preorder(node):
    if node:
        print(node.data, end=' ')       # menampilkan node utama
        preorder(node.left)             # menampilkan node kiri
        preorder(node.right)            # menampilkan node kanan

# Traversal postorder
def postorder(node):
    if node:
        postorder(node.left)            # menampilkan node kiri
        postorder(node.right)           # menampilkan node kanan
        print(node.data, end=' ')       # menampilkan node utama

print("Hasil inorder traversal:")
inorder(akar)

print("\nPreorder Traversal:")
preorder(akar)

print("\nPostorder Traversal:")
postorder(akar)

print("\n", "=" * 40)
print("Pencarian key:")

# Fungsi pencarian data
def cari(node, key):
    if node is None or node.data == key:
        return node     # mengembalikan node jika data ditemukan

    if key < node.data:
        return cari(node.left, key)    # mencari ke node kiri

    return cari(node.right, key)       # mencari ke node kanan

# Uji pencarian pertama
key = 40
hasil = cari(akar, key)

if hasil:
    print(f"{key} ditemukan dalam tree.")
else:
    print(f"{key} tidak ditemukan.")

# Uji pencarian kedua
key = 45
hasil = cari(akar, key)

if hasil:
    print(f"{key} ditemukan dalam tree.")
else:
    print(f"{key} tidak ditemukan.")
```
