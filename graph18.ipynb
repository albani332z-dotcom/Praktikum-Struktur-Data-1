```python
# Penugasan Graph Python

# Membuat graph tak berarah
graph = {
    'K': ['L', 'M', 'N'],
    'L': ['K', 'M', 'O'],
    'M': ['K', 'L', 'N', 'O'],
    'N': ['K', 'M', 'O'],
    'O': ['L', 'M', 'N']
}

# Fungsi menampilkan graph
def print_graph(graph):
    for vertex in graph:
        print(vertex, "terhubung ke", graph[vertex])

# BFS
def bfs(graph, start):
    visited = []
    queue = [start]

    while queue:
        current = queue.pop(0)

        if current not in visited:
            visited.append(current)

            for neighbor in graph[current]:
                if neighbor not in visited:
                    queue.append(neighbor)

    return visited

# DFS
def dfs(graph, start, visited=None):
    if visited is None:
        visited = []

    visited.append(start)

    for neighbor in graph[start]:
        if neighbor not in visited:
            dfs(graph, neighbor, visited)

    return visited

# Fungsi mencari jalur
def find_path(graph, start, end, path=None):
    if path is None:
        path = []

    path = path + [start]

    if start == end:
        return path

    for neighbor in graph[start]:
        if neighbor not in path:
            result = find_path(graph, neighbor, end, path)

            if result:
                return result

    return None

# Fungsi mengecek keterhubungan graph
def is_connected(graph):
    start_node = list(graph.keys())[0]
    visited_nodes = dfs(graph, start_node)

    return len(visited_nodes) == len(graph)

# Menampilkan hasil
print("Daftar Graph")
print_graph(graph)

print("\nHasil BFS :", bfs(graph, 'K'))
print("Hasil DFS :", dfs(graph, 'K'))
print("Jalur K ke O :", find_path(graph, 'K', 'O'))
print("Graph Terhubung :", is_connected(graph))
```
